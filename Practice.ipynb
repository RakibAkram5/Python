```python
def factorial(number):
    result = 1
    for i in range(1, number + 1):
        result = result * i
    return result  

number = int(input("Enter the Number: "))
print(f"The Factorial of {number}! is: {factorial(number)}")
```

```python
def my_len(sequence):
    count = 0
    for _ in sequence:
        count += 1
    return count

def count_pattern(pattern, lst):
    count = 0 
    pattern_length = my_len(pattern)
    
    for i in range(my_len(lst) - pattern_length + 1):
        current_slice = tuple(lst[i:i + pattern_length])
        if current_slice == pattern:
            count += 1
    return count   


print(count_pattern(('a', 'b'), ('a','b', 'c', 'e', 'b', 'a', 'b', 'f')))  
print(count_pattern(('a', 'b', 'a'), ('g', 'a', 'b', 'a', 'b', 'a','b', 'a')))
```

```python
def addition(num1, num2):
    return num1 + num2

def subtraction(num1, num2):
    return num1 - num2

def multiplication(num1, num2):
    return num1 * num2

def division(num1, num2):
    if num2 == 0:
        return "Error! Division by zero."
    return num1 / num2

print("Simple Calculator")
num1 = float(input("Enter the Number 1: "))
num2 = float(input("Enter the Number 2: "))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your Choice: "))

if choice == 1:
    print("Result:", addition(num1, num2))
elif choice == 2:
    print("Result:", subtraction(num1, num2))
elif choice == 3:
    print("Result:", multiplication(num1, num2))
elif choice == 4:
    print("Result:", division(num1, num2))
else:
    print("Invalid choice!")
```

```python
def sortingAlphabets(sentence):
    sentence += " "
    words = []
    word = ""
    for i in range(len(sentence)):
        if sentence[i] != " ":
            word += sentence[i]
        else:
            if word != "":
                words += [word]
                word = ""
    for i in range(len(words)):
        for j in range(i + 1, len(words)):
            if words[i].lower() > words[j].lower():
                words[i], words[j] = words[j], words[i]
    print("Sorted Sentence:", end=" ")
    for i in range(len(words)):
        print(words[i][:], end=" ")

sentence = input("Enter a sentence: ")
sortingAlphabets(sentence)
```

```python
class IntegerToRoman:
    def __init__(self):
        self.values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
        self.symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL", "X", "IX", "V", "IV", "I"]

    def convert(self, number):
        roman = ""
        i = 0
        while number > 0:
            check = number // self.values[i]
            for _ in range(check):
                roman += self.symbols[i]
                number -= self.values[i]
            i += 1
        return roman


# Example usage:
obj = IntegerToRoman()
number=int(input("Enter the Number to Convert in Roman: "))
print(f"Roman Number: {obj.convert(number)}")
```

```python
class PairFinder:
    def __init__(self, numbers, target):
        self.numbers = numbers
        self.target = target

    def find_pair(self):
        for i in range(len(self.numbers)):
            for j in range(i + 1, len(self.numbers)):
                sum = self.numbers[i] + self.numbers[j]
                if sum == self.target:
                    return i + 1, j + 1 
        return None


numbers = [10, 20, 30, 40, 50]
target = int(input("Enter the Number: "))
obj = PairFinder(numbers, target)
print(f"Indices: {obj.find_pair()}")
```

```python
class ThreeSum:
    def __init__(self, numbers):
        self.numbers = numbers

    def find_triplets(self):
        triplets = []
        nums = sorted(self.numbers)
        n = len(nums)

        for i in range(n - 2):
            if i > 0 and nums[i] == nums[i - 1]:
                continue  # skip duplicates

            left, right = i + 1, n - 1
            while left < right:
                total = nums[i] + nums[left] + nums[right]
                if total == 0:
                    triplets.append([nums[i], nums[left], nums[right]])
                    left += 1
                    right -= 1
                    while left < right and nums[left] == nums[left - 1]:
                        left += 1
                    while left < right and nums[right] == nums[right + 1]:
                        right -= 1
                elif total < 0:
                    left += 1
                else:
                    right -= 1

        return triplets



numbers = [-25, -10, -7, -3, 2, 4, 8, 10]
obj = ThreeSum(numbers)
print(obj.find_triplets())
```

```python
class StringReverse:
    def __init__(self, text):
        self.text = text

    def reverse_word(self):
        text = self.text + " "
        word = ""
        words = []
        for ch in text:
            if ch != " ":
                word += ch
            else:
                if word != "":
                    words.append(word)
                    word = ""

        reversed_text = ""
        for i in range(len(words) - 1, -1, -1):
            reversed_text += words[i]
            if i != 0:
                reversed_text += " "
        return reversed_text


# Example usage
s = StringReverse("hello .py")
print(s.reverse_word())
```

```python
def calculator(number1, number2, operation='addition', output_format='float'):
    
    try:
        number1 = float(number1)
        number2 = float(number2)
    except:
        raise TypeError("Both number1 and number2 must be numbers (int or float).")

   
    valid_operations = ['addition', 'subtraction', 'multiplication', 'division']
    if operation not in valid_operations:
        raise ValueError(f"Invalid operation. Choose from {valid_operations}")

    
    valid_formats = ['integer', 'float']
    if output_format not in valid_formats:
        raise ValueError(f"Invalid output format. Choose from {valid_formats}")

   
    if operation == 'addition':
        result = number1 + number2
    elif operation == 'subtraction':
        result = number1 - number2
    elif operation == 'multiplication':
        result = number1 * number2
    elif operation == 'division':
        if number2 == 0:
            raise ZeroDivisionError("Division by zero is not allowed.")
        result = number1 / number2

    
    if output_format == 'integer':
        return round(result)
    else:
        return float(result)

print(calculator(10, 5))                       
print(calculator(10, 5, 'subtraction'))         
print(calculator(10, 5, 'multiplication'))      
print(calculator(10, 3, 'division', 'integer')) 
print(calculator(10, 3, 'division', 'float'))   
```

```python
def find_mean_median_mode():
    user_input = input("Enter numbers separated by space: ")
    elements = user_input.split()
    numbers = []
    for item in elements:
        numbers.append(float(item))
    total = 0
    for num in numbers:
        total += num
    count = len(numbers)
    mean = total / count
    numbers.sort()
    if count % 2 == 0:
        median = (numbers[count // 2 - 1] + numbers[count // 2]) / 2
    else:
        median = numbers[count // 2]
    frequency = {}
    for num in numbers:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1
    max_freq = 0
    for value in frequency.values():
        if value > max_freq:
            max_freq = value
    mode = []
    for num, freq in frequency.items():
        if freq == max_freq:
            mode.append(num)
    print("\nResults:")
    print(f"Sum: {total}")
    print(f"Mean: {mean}")
    print(f"Median: {median}")
    print(f"Mode: {mode}")

find_mean_median_mode()
```

```python
def dups(numbers):
    duplicates = []
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] == numbers[j]:
                if numbers[i] not in duplicates:
                    duplicates.append(numbers[i])
    return duplicates

def main():
    elements = []
    n = int(input("Enter number of elements: "))
    for i in range(n):
        item = input("Enter element " + str(i + 1) + ": ")
        elements.append(item)
    result = dups(elements)
    print("Duplicate elements are:", result)

main()
```
